# OQIM AI - Sodda va Tezkor Versiya
Ushbu kod og'ir xaritalarsiz, faqat AI algoritmlari (K-Means va Q-Learning) qanday ishlashini soniyalarda ko'rsatib beradi.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import random
import time

print("✅ Kutubxonalar yuklandi!")
```

## 1. Tirbandlikni Klasterlash (K-Means)
Shahar chorrahalarida tasodifiy mashinalar yig'ilishi (probka) hosil qilamiz va K-Means orqali 3 ta xavf darajasiga bo'lamiz.

```python
# Tasodifiy 200 ta mashina lokatsiyasini (X, Y koordinatalarda) yaratish
np.random.seed(42)
x_coords = np.random.normal(50, 15, 200) # Shahar markazi atrofida
y_coords = np.random.normal(50, 15, 200)
vehicles = np.column_stack((x_coords, y_coords))

# K-Means algoritmini ishga tushirish (3 ta klaster)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(vehicles)
centers = kmeans.cluster_centers_

# Natijani chizish
plt.figure(figsize=(8, 6))
colors = ['green', 'orange', 'red']
for i in range(3):
    plt.scatter(vehicles[labels == i, 0], vehicles[labels == i, 1], c=colors[i], label=f'{i}-Klaster')
    plt.scatter(centers[i, 0], centers[i, 1], c='black', marker='X', s=200)

plt.title("OQIM AI: Tirbandlik zonalarini aniqlash (K-Means)")
plt.legend()
plt.grid(True)
plt.show()

print("🟢 Yashil - Erkin, 🟠 Sariq - O'rtacha, 🔴 Qizil - Og'ir tirbandlik")
```

## 2. Svetaforlarni O'qitish (Q-Learning)
Kutayotgan mashinalar soniga qarab qancha vaqt yashil chiroq yoqish kerakligini AI o'zi o'rganadi.

```python
class TrafficAI:
    def __init__(self):
        self.actions = [15, 30, 45, 60] # Yashil chiroq soniyalari
        self.q_table = {}
        
    def get_state(self, queue_len):
        # Har 5 ta mashina bitta holat deb olinadi
        return min(queue_len // 5, 5)
        
    def train(self, episodes=1000):
        print(f"Q-Learning {episodes} marta mashq qilmoqda...")
        for _ in range(episodes):
            queue = random.randint(0, 30) # 0 dan 30 gacha mashina kutyapti
            state = self.get_state(queue)
            
            if state not in self.q_table:
                self.q_table[state] = {a: 0.0 for a in self.actions}
                
            # Tasodifiy harakat tanlash
            action = random.choice(self.actions)
            
            # Mukofot (Reward) hisoblash. Qancha kam mashina qolsa shuncha yaxshi
            cars_passed = action / 2
            remaining = max(0, queue - cars_passed)
            reward = -remaining
            
            # Jadvalni yangilash
            self.q_table[state][action] = self.q_table[state][action] + 0.1 * (reward - self.q_table[state][action])
        print("✅ AI o'qitildi!\n")

    def predict(self, queue):
        state = self.get_state(queue)
        best_time = max(self.q_table.get(state, {30: 0}), key=self.q_table.get(state, {30: 0}).get)
        return best_time

# Modelni yaratish va o'qitish
ai_signal = TrafficAI()
ai_signal.train(2000)

# Natijalarni sinash
test_cases = [2, 10, 18, 28]
for q in test_cases:
    time.sleep(0.5)
    green_time = ai_signal.predict(q)
    print(f"🚦 Chorrahada {q} ta mashina kutyapti -> AI yashil chiroqni {green_time} soniyaga yoqdi!")
```
